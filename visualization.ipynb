```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("data.csv")
```

```python
age_counts = df["What is your age?"].value_counts()
plt.figure(figsize=(6,4))
plt.bar(age_counts.index, age_counts.values)
plt.title("Distribution of Age")
plt.xlabel("Age Groups")
plt.ylabel("Number of Students")
plt.show()
```

```python
age_counts = df["What is your age?"].value_counts()
plt.pie(
    age_counts,
    labels=age_counts.index,
    autopct="%1.1f%%"
)
plt.title("Distribution of Age")
plt.show()
```

```python
screen_time_counts = df["How many hours do you spend on screens daily? (including phone, laptop, tablets )"].value_counts()

plt.bar(screen_time_counts.index, screen_time_counts.values, color="khaki")
plt.title("Daily Screen Time")
plt.xlabel("Hours")
plt.xticks(rotation=45)
plt.ylabel("Count")
plt.show()
```

```python
coffee_tea_counts = df["Do you prefer coffee or tea?"].value_counts()

plt.pie(
    coffee_tea_counts,
    labels=coffee_tea_counts.index,
    autopct="%1.1f%%",
    colors=["orange", "skyblue", "teal"]
)

plt.title("Coffee or Tea Preference")
plt.show()
```

```python
study_time_counts = df["When do you prefer to study?"].value_counts()

plt.bar(
    study_time_counts.index,
    study_time_counts.values
)

plt.title("Preferred Study Time")
plt.xlabel("Time of Day")
plt.ylabel("Count")
plt.show()
```

```python
social_media_counts = df["Which social media app do you prefer?"].value_counts()

plt.bar(
    social_media_counts.index,
    social_media_counts.values
)

plt.title("Social Media App Preferences")
plt.xlabel("Social Media App Categories")
plt.ylabel("Number of Students")
plt.xticks(rotation=45)
plt.show()
```

```python
hobbies_media_counts = df["What are your hobbies / interest? "].value_counts()

plt.barh(
    hobbies_media_counts.index,
    hobbies_media_counts.values
)

plt.title("Hobbies / Interest")
plt.xlabel("Number of Students")
plt.ylabel("Hobbies")
plt.show()
```

```python
language_counts = df["How many languages do you speak fluently?"].value_counts()

plt.bar(
    language_counts.index,
    language_counts.values
)

plt.title("Distribution of Number of Languages Spoken")
plt.xlabel("Number of Languages")
plt.ylabel("Number of Students")
plt.show()
```

```python
exercise_counts = df["How often do you exercise in a week?"].value_counts()

plt.bar(
    exercise_counts.index,
    exercise_counts.values
)

plt.title("Exercise Frequency per Week")
plt.xlabel("Frequency")
plt.xticks(rotation=45)
plt.ylabel("Count")
plt.show()
```

```python
Commute_counts = df["How do you primarily commute to work or school?"].value_counts()

plt.bar(Commute_counts.index, Commute_counts.values)
plt.title("Primary Commute Method Distribution")
plt.xlabel("Commute Method")
plt.xticks(rotation=45)
plt.ylabel("Count")
plt.show()
```

```python
Sleep_hours_counts = df["How many hours of sleep do you get on an average night?"].value_counts()
plt.bar(Sleep_hours_counts.index, Sleep_hours_counts.values)
plt.title("Average Nightly Sleep Hours")
plt.xlabel("Hours")
plt.xticks(rotation=45)
plt.ylabel("Count")
plt.show()
```

```python
source_counts = df["What is your preferred method for staying informed about current events?"].value_counts()

plt.bar(
    source_counts.index,
    source_counts.values
)
plt.title("Preferred Current Events Source")
plt.xlabel("Source")
plt.xticks(rotation=45)
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(
    x="What is your age?",
    hue="When do you prefer to study?",
    data=df
)
plt.title("Age vs. Preferred Study Time")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(
    x='What is your age?',
    hue='How often do you exercise in a week?',
    data=df
)
plt.title('Age vs. Exercise Frequency')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.countplot(
    x='How many hours do you spend on screens daily? (including phone, laptop, tablets )',
    hue='What is your age?',
    data=df
)
plt.title('Daily Screen Time vs. Age')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(
    x='How many languages do you speak fluently?',
    hue='Do you prefer coffee or tea?',
    data=df
)
plt.title('Coffee or Tea Preference vs. Number of Languages Spoken')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.countplot(
    x='How do you primarily commute to work or school?',
    hue='How often do you exercise in a week?',
    data=df,
    width=0.6
)
plt.title('Commute Method vs. Exercise Frequency')
plt.xlabel('Commute Method')
plt.ylabel('Count')
plt.legend(title='Exercise Frequency', fontsize='small')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.countplot(
    x='How many hours of sleep do you get on an average night?',
    hue='How many hours do you spend on screens daily? (including phone, laptop, tablets )',
    data=df,
    width=0.7
)
plt.title('Sleep Hours vs. Daily Screen Time')
plt.xlabel('Sleep Hours')
plt.ylabel('Count')
plt.legend(title='Screen Time', loc='upper right')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
